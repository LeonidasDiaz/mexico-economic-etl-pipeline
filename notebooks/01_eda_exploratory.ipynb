# 01. Exploratory Data Analysis (EDA) - Banxico API Integration

## 1. Project Objective
Extract and explore macroeconomic indicators from the official Banxico SIE API to validate data types, check for missing values, and prepare the schema for the ETL pipeline.

---

## 2. Environment Setup & API Authentication
Loading environment variables from the `.env` file to securely access the Banxico API Token without hardcoding sensitive credentials.

```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
from datetime import datetime

# Cargar el archivo .env especificando la ruta desde la raiz
load_dotenv(dotenv_path = "../.env")

# Leer el token del archivo .env
BANXICO_TOKEN = os.getenv("BANXICO_TOKEN")
# Parámetros de la API
series_id = "SF63528"# Tipo de Cambio FIX (Pesos por dólar)
fecha_inicio = "2020-01-01"
fecha_fin = datetime.now().strftime("%Y-%m-%d")

url = f"https://www.banxico.org.mx/SieAPIRest/service/v1/series/{series_id}/datos/{fecha_inicio}/{fecha_fin}"
headers = {"Bmx-Token": BANXICO_TOKEN}

# Petición a la API
response = requests.get(url, headers=headers)

# Validación de respuesta HTTP y tipo de contenido JSON
if response.status_code == 200 and "application/json" in response.headers.get("Content-Type", ""):
    raw_data = response.json()['bmx']['series'][0]['datos']
    
    # Cargar datos en Pandas
    df = pd.DataFrame(raw_data)
    print("Conexión exitosa. Primeros 5 registros:")
    print(df.head())
else:
    print(f"Error en la conexión. Status Code: {response.status_code}")
    print(response.text)
```
